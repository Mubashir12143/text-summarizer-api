```python
import gradio as gr
from summarizer import summarize_text

def generate_summary(text, language):
    return summarize_text(text, language)

interface = gr.Interface(
    fn=generate_summary,
    inputs=[
        gr.Textbox(lines=8, label="Enter your text"),
        gr.Radio(["english", "urdu"], label="Select Language")
    ],
    outputs=gr.Textbox(lines=5, label="Summary"),
    title="Text Summarization App (English + Urdu)",
    description="Summaries long text into short meaningful text."
)

interface.launch()
```
